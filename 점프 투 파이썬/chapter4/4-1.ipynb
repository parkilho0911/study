### 매개변수를 지정하여 호출하기

```python
def sub(a, b):
    return a - b

result = sub(a=7, b=3)  # a에 7, b에 3을 전달
print(result)
```

```python
def sub(a, b):
    return a - b

result = sub(b=7, a=3)  # 위치를 바꿔도 매개변수에는 잘 들어간다
print(result)
```

```python
def add_many(*args): 
    result = 0 
    for i in args: 
       result = result + i   # *args에 입력받은 모든 값을 더한다.
    return result 
    
print(add_many(1,2,3,4,5))
```

```python
def add_mul(choice, *args): 
    if choice == "add":   # 매개변수 choice에 "add"를 입력받았을 때
        result = 0 
        for i in args: 
            result = result + i 
    elif choice == "mul":   # 매개변수 choice에 "mul"을 입력받았을 때
        result = 1 
        for i in args: 
            result = result * i 
    return result 

result = add_mul("add", 1,2,3,4)
print(result)
```

### 특별한 상황일 때 함수를 빠져나가고 싶다면 return을 단독으로 써서 함수를 즉시 빠져나갈 수 있다. 다음 예를 살펴보자.

```python
def say_nick(nick): 
    if nick == "바보": 
        return 
    print("나의 별명은 %s 입니다." % nick)

say_nick("바보")
say_nick("하이")

#이렇게 바보라고 입력이 되면 문자열을 출력하지 않고 즉시 함수를 빠져나간다.
```

### 매개변수에 초깃값 미리 설정하기

```python
# default1.py
def say_myself(name, age, man=True): #기본값을 지정해주는 값들은 맨뒤로 보내줘야한다.
    print("나의 이름은 %s 입니다." % name) 
    print("나이는 %d살입니다." % age) 
    if man: 
        print("남자입니다.")
    else: 
        print("여자입니다.")

say_myself("일호", 25)
say_myself("세희", 22, False) 
#기본값이 man으로 설정이 되어 있기 때문에 False를 넣어줘야한다.
```

위 오류 메시지는 ‘초깃값이 없는 매개변수(age)는 초깃값이 있는 매개변수(man) 뒤에 사용할 수 없다’라는 뜻이다. 즉, 매개변수로 (name, age, man=True)는 되지만, (name, man=True, age)는 안 된다는 것이다. 초기화하고 싶은 매개변수는 항상 뒤쪽에 놓아야 한다는 것을 잊지 말자.

### 함수 안에서 선언한 변수의 효력 범위

```python
# vartest.py
a = 1
def vartest(a):
    a = a +1 #함수안에 들어있는 변수는 지역변수이다.
             #함수가 끝나면 지역변수는 더 이상 사용되지 않는다.
vartest(a)
print(a)  #여기서 출력되는 값은 함수 밖에 있는 전역 변수이다.
```

```python
# vartest_return.py
a = 1 
def vartest(a): 
    a = a +1  
    return a  #리턴값을 통해서 vartest에 저장이 된다.

a = vartest(a) #그 저장된 값을 a에 넣어준다. 만약 vartest의 값을 넣어줄 곳이 없다면
               #지역 변수의 값이 아닌 전역 변수의 값으로 출력 된다.
print(a) #그러면 a의 값은 vartest와 동일한 값이 출력이 된다
```

```python
# vartest_global.py
a = 1 
def vartest(): 
    global a # 전역변수 a를 가져온다.
    a = a+1 #전역변수를 가져온 값에서 플러스한다.

vartest() 
print(a)
```

```python
b = [1,2,3]
def vartest2(b): #여기서 b는 위의 전역변수 리스트를 가져온 것이다.
    b = b.append(4) #그래서 append를 사용하면 전역변수 리스트에 추가 된 것이 맞다.
vartest2(b)         #b=None값이 된다 리턴값이 없기 때문에
print(b)
```

```python
a = "hello" #전역변수 
def vartest(a): #지역변수 이뮤터블형은 값이 변하지 않기 때문에 지역변수에서 새롭게 만든 값이 된다.
    a = a + 'hi' #그래서 a의 값은 변하지 않고, 새롭게 생성되어 있다. 
vartest(a)
print(a)
```

```python
a = "hello" #전역변수 
def vartest(a): #이뮤터블형 이라서 값이 변하지 않고 새롭게 생성이 된다.
    a = a + 'hi' # hello에다가 hi를 더한 값을 생성한다 vartest(a)로
    return a  #새로 생성된 값을 a에 돌려준다.
a = vartest(a)
print(a) #출력하면 새롭게 생성된 값이 출력 된다.
```
