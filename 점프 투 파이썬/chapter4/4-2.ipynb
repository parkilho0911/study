### 큰따옴표로 둘러싸인 문자열은 + 연산과 동일하다

```python
print("life" "is" "too short")  # 1번
print("life"+"is"+"too short")  # 2번
```

문자열 띄어쓰기는 쉼표로 한다

```python
print("life", "is", "too short")
```

한 줄에 결괏값 출력하기

```python
for i in range(10):
    print(i, end=' ')
#end 매개변수의 초깃값은 줄바꿈(\n) 문자이다.
```
