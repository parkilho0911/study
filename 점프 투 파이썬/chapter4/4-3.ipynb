```python
# newfile.py
f = open("새파일.txt", 'w')
f.close()
```


파일열기모드	설명
r	읽기 모드: 파일을 읽기만 할 때 사용한다.
w	쓰기 모드: 파일에 내용을 쓸 때 사용한다.
a	추가 모드: 파일의 마지막에 새로운 내용을 추가할 때 사용한다.
간혹 한글 파일이 깨져서 보일때가 있는데 그때는 인코딩을 해줘야한다
encoding = "UTF-8" <<이것을 사용하면 한글이 꺠지지 않는다.

```python
# write_data.py
f = open("c:\\Users\\ilho\\Desktop\\점프 투 파이썬\\chapter4\\새파일.txt", 'w', encoding = "UTF-8")
for i in range(1, 11):
    data = "%d번째 줄입니다.\n" % i
    f.write(data)
f.close()
```

```python
# UTF-8 인코딩을 명시적으로 지정
f = open("C:\\Users\\ilho\\Desktop\\점프 투 파이썬\\chapter4\\새파일.txt", 'r', encoding="utf-8")
line = f.readline()
print(line)
f.close()
```

```python
# readline_all.py
f = open("C:\\Users\\ilho\\Desktop\\점프 투 파이썬\\chapter4\\새파일.txt", 'r', encoding="utf-8")
while True:
    line = f.readline()
    if not line: break
    print(line)
f.close()
```

```python
# readlines.py
f = open("C:\\Users\\ilho\\Desktop\\점프 투 파이썬\\chapter4\\새파일.txt", 'r', encoding="utf-8")
lines = f.readlines()
for line in lines:
    print(line)
f.close()
```

```python
f = open("C:\\Users\\ilho\\Desktop\\점프 투 파이썬\\chapter4\\새파일.txt", 'r', encoding="utf-8")
lines = f.readlines()
for line in lines:
    line = line.strip()  # 줄 끝의 줄 바꿈 문자를 제거한다.
    #line.replace('\n, ' ')로 바꾸어서 사용가능 replace가 변경하는 것이기 떄문에에
    print(line)
f.close()
```

### read 함수 사용하기

### 파일 객체를 for 문과 함께 사용하기

```python
# read_for.py
f = open("C:\\Users\\ilho\\Desktop\\점프 투 파이썬\\chapter4\\새파일.txt", 'r', encoding="utf-8")
for line in f:
    print(line)
f.close()
```

### 파일에 새로운 내용 추가하기

```python
# add_data.py
f = open("C:\\Users\\ilho\\Desktop\\점프 투 파이썬\\chapter4\\새파일.txt", 'a', encoding="utf-8")
for i in range(11, 20):
    data = "%d번째 줄입니다.\n" % i
    f.write(data)
f.close()
```

### with 문과 함께 사용하기

```python
f = open("foo.txt", 'w')
f.write("Life is too short, you need python")
f.close()
```

```python
# file_with.py
with open("foo.txt", "w") as f:
    f.write("Life is too short, you need python")
```
