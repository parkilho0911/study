### 문자열 소비가 없는 메타 문자

아직 살펴보지 않은 메타 문자에 대해서 모두 살펴보자. 여기에서 다룰 메타 문자는 앞에서 살펴본 메타 문자와 성격이 조금 다르다. 앞에서 살펴본 +, *, [], {} 등의 메타 문자는 매치가 성사되면 문자열을 탐색하는 시작 위치가 변경된다(보통 소비된다고 표현한다). 가령 aac라는 문자열에서 a+라는 패턴을 찾아야 할 때, aa가 매치되고 나면 문자열 중 aa는 소비되고 남은 c가 시작 위치가 된다.

```python
import re
p = re.compile('Crow|Servo')
m = p.match('CrowHello')
print(m)
```

### ^

^ 메타 문자는 문자열의 맨 처음과 일치한다는 것을 의미한다. 앞에서 살펴본 컴파일 옵션 re.MULTILINE을 사용할 경우에는 여러 줄의 문자열일 때 각 줄의 처음과 일치하게 된다.<br>
^를 사용할 떄에는 ^뒤의 문자열로 시작을 해야 매칭이 된다.

```python
import re
print(re.search('^Life', 'Life is too short'))
print(re.search('^Life', 'My Life'))
```

### $

$ 메타 문자는 ^ 메타 문자와 반대의 경우이다. 즉, $는 문자열의 끝과 매치한다는 것을 의미한다.

```python
import re
print(re.search('short$', 'Life is too short'))
print(re.search('My$', 'My Life'))
```

### \A

\A는 문자열의 처음과 매치된다는 것을 의미한다. ^ 메타 문자와 동일한 의미이지만, re.MULTILINE 옵션을 사용할 경우에는 다르게 해석된다. re.MULTILINE 옵션을 사용할 경우 ^은 각 줄의 문자열의 처음과 매치되지만, <br>\A는 줄과 상관없이 전체 문자열의 처음하고만 매치된다.

### \Z

\Z는 문자열의 끝과 매치된다는 것을 의미한다. 이것 역시 \A와 동일하게 re.MULTILINE 옵션을 사용할 경우, $ 메타 문자와는 달리 전체 문자열의 끝과 매치된다.

### \b

```python
import re
p = re.compile(r'\bclass\b')
print(p.search('no class at all'))  
```

### \B

\B 메타 문자는 \b 메타 문자와 반대의 경우이다. 즉, 화이트스페이스로 구분된 단어가 아닌 경우에만 매치된다.<br>
class 단어의 앞뒤에 화이트스페이스가 하나라도 있는 경우에는 매치가 되지 않는다는 것을 확인할 수 있다.

```python
import re
p = re.compile(r'\Bclass\B')
print(p.search('no class at all')) 
print(p.search('the declassified algorithm'))
print(p.search('one subclass is'))
```

### 그루핑

```python
import re
p = re.compile('(ABC)+')
m = p.search('ABCABCABC OK?')
print(m)
```

```python
import re
p = re.compile(r"\w+\s+\d+[-]\d+[-]\d+")
m = p.search("park 010-1234-1234")
print(m)
```

```python
import re
p = re.compile(r"(\w+)+\s+\d+[-]\d+[-]\d+")
m = p.search("park 010-1234-1234")
print(m.group(1))
```

```python
import re
p = re.compile(r"\w+\s+(\d+[-]\d+[-]\d+)+")
m = p.search("park 010-1234-1234")
print(m.group(1))
```

앞 예에서 볼 수 있듯이 (\w+)\s+((\d+)[-]\d+[-]\d+)처럼 그룹을 중첩해 사용할 수도 있다. 그룹이 중첩된 경우는 바깥쪽부터 시작해 안쪽으로 들어갈수록 인덱스 값이 증가한다.

### 그루핑된 문자열 재참조하기

```python
import re
p = re.compile(r'(\b\w+)\s+\1')
print(p.search('Paris in the the spring').group())
```

### 그루핑된 문자열에 이름 붙이기

(?P<name>\w+)\s+((\d+)[-]\d+[-]\d+)

```python
import re
p = re.compile(r"(?P<name>\w+)\s+((\d+)[-]\d+[-]\d+)")
m = p.search("park 010-1234-1234")
print(m.group("name"))
```

### 전방 탐색

```python
import re
p = re.compile(".+:")
m = p.search("http://google.com")
print(m.group())
```

전방 탐색에는 긍정(positive)과 부정(negative)의 2가지 종류가 있고 다음과 같이 표현한다.

긍정형 전방 탐색((?=...)): ...에 해당하는 정규식과 매치되어야 하며 조건이 통과되어도 문자열이 소비되지 않는다.<br>
부정형 전방 탐색((?!...)): ...에 해당하는 정규식과 매치되지 않아야 하며 조건이 통과되어도 문자열이 소비되지 않는다.

```python
p = re.compile(".+(?=:)")
m = p.search("http://google.com")
print(m.group())
```

### 문자열 바꾸기

```python
import re
p = re.compile('(blue|white|red)')
print(p.sub('colour', 'blue socks and red shoes'))
```

### sub 메서드 사용 시 참조 구문 사용하기

```python
import re
p = re.compile(r"(?P<name>\w+)\s+(?P<phone>(\d+)[-]\d+[-]\d+)")
print(p.sub("\g<phone> \g<name>", "park 010-1234-1234"))
```

### sub 메서드의 매개변수로 함수 넣기

```python
import re
def hexrepl(match):
    value = int(match.group())
    return hex(value)
p = re.compile(r'\d+')
p.sub(hexrepl, 'Call 65490 for printing, 49152 for user code.')
```
