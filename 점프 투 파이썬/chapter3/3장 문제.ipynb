while문을 사용해 1부터 1000까지의 자연수 중 3의 배수의 합을 구해보자자

```python
result = 0
i = 1
while i <= 1000:
    if i % 3 ==0:
        result +=i
    i += 1

print(result)
```

while 문을 사용하여 다음과 같이 별(*)을 표시하는 프로그램을 작성해보자자

**

A학급에 총 10명의 학생이 있다. 이 학생들의 중간고사 점수는 다음과 같다
[70,60,55,75,95,90,80,80,85,100]

```python
from numpy import average


A = [70,60,55,75,95,90,80,80,85,100]
total = 0
for score in A:
    total +=score
average = total/len(A) #리스트의 숫자의 갯수는 len을 이용하면 된다.
print(average)
```

리스트 중에서 홀수에만 2를 곱하여 저장하는 다음 코드가 있다.

```python
numbers = [1,2,3,4,5]
result = []
for n in numbers:
    if n % 2 == 1:
        result.append(n*2)
```

```python
numbers = [1,2,3,4,5]
result = [n*2 for n in numbers if n % 2 ==1]
print(result)
#리스트 내포를 이용할 때는 굳이 append를 사용하지 않고 뒤에 값들만 이용한다.
```
