```python
a = b = [1,2,3]
a[1] = 4
print(b)
# 정답은 [1,4,3]이 나오고 이유는 a하고b는 동일한 리스트를 말하고 있다 그러니 a[1]을 4로 바꾼다 한들 a만 변경되는 것이 아닌 b까지 함께 변경이 된다.
```
