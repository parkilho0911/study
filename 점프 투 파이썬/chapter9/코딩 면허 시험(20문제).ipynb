1. 문자열 바꾸기 ★☆☆

a:b:c:d<br>
문자열의 split와 join함수를 사용하여 위 문자열을 다음과 같이 고치시오

```python
a = "a:b:c:d"
b = a.split(":") # :를 기준으로 나눈다. 그러면 split()에서 사용한 :는 삭제가 된다다
c = "#".join(b)  # 각 리스트의 요소에 #를 추가하게 된다. 그렇게 되면 리스트와 추가한 것이 이어 붙게 된다.
print(c)
```

2. 딕셔너리 값 추출하기 ★☆☆<br>
다음은 딕셔너리의 a에서 'C'라는 key에 해당하는 value를 출력하는 프로그램이다.

```python
a = {'A':90, 'B':80}
a['C']
```

a 딕셔너리에 'C'라는 key가 없으므로 위와 같은 오류가 발생한다. 'C'에 해당하는 key값이 없을 경우 오류 대신 70을 얻을 수 있도록 수정하시오.

```python
a = {'A':90, 'B':80, 'C':1}
a.get('C',70)
```

3. 리스트의 더하기와 extend 함수 ★★☆

```python
a=[1,2,3]
a = a + [4,5]
a
```

```python
a=[1,2,3]
a.extend([4,5])
a
```

### +기호를 사용하여 더한 것과 extend한 것의 차이점이 있을까? 있다면 그 차이점을 설명하시오 <br>
+를 사용하는 것은 원래 있는 값에서 추가를 하여 다른 값으로 변경이 되지만 <br> extend함수를 사용하는 것은 그 리스트를 확장하는 것이기 때문에 값이 변하지 않고 추가하는 것이 확장 된다는 차이점이 있다다

5. 피보나치 함수 ★★★

첫 번쨰 항의 값이 0이고 두 번쨰 항의 값이 1일 때, 이후에 이어지는 항은 이전의 두 항을 더한 값으로 이루어지는 수열을 피보나치 수열이라고 한다. <br>
0,1,1,2,3,5,8,13 ... <br>
입력을 정부 n으로 받았을 때, n이하까지의 피보나치 수열을 출력하는 함수를 작성해보자.

```python
n = int(input())

def fibo(n):
    if n == 0: 
        return 0
    if n == 1:
        return 1
    if n >= 2:
        return fibo(n-1) + fibo(n-2)
    
for i in range(n):
    print(fibo(i),end=" ")
```
